```python
import kaleidocell
import scanpy as sc
import os
```

```python
adata = sc.read_h5ad("/net/bq-storage/ag-cherrmann/bq_cinac/projects/synimmune/data/01_extGBmap/immune_cells_extended_gbmap.h5ad")

```

```python
print(adata.obs["annotation_level_3"].value_counts())
```

```python
# Create subsets
cell_types = adata.obs["annotation_level_3"].unique()

subsets = {}
for cell_type in cell_types:
    key = cell_type.replace("/", "_").replace(" ", "_")
    subsets[key] = adata[adata.obs["annotation_level_3"] == cell_type].copy()
    print(f"{cell_type}: {subsets[key].n_obs} cells")
```

```python
# Save subsets
output_dir = "/net/bq-storage/ag-cherrmann/bq_cinac/projects/synimmune/data/immune_subsets/"
os.makedirs(output_dir, exist_ok=True)

for name, subset in subsets.items():
    path = os.path.join(output_dir, f"{name}.h5ad")
    subset.write_h5ad(path)
    print(f"Saved {name} to {path}")
```
