```python
import kaleidocell
import scanpy as sc
import os
```

```python
def filter_donors(adata, min_cells=50):
    donor_counts = adata.obs["donor_id"].value_counts()
    valid_donors = donor_counts[donor_counts >= min_cells].index
    adata_filtered = adata[adata.obs["donor_id"].isin(valid_donors)].copy()
    print(f"Donors before: {adata.obs['donor_id'].nunique()}")
    print(f"Donors after: {adata_filtered.obs['donor_id'].nunique()}")
    print(f"Cells before: {adata.n_obs}")
    print(f"Cells after: {adata_filtered.n_obs}")
    return adata_filtered

# Load and filter each subset
cd4_cd8 = sc.read_h5ad("/net/bq-storage/ag-cherrmann/bq_cinac/projects/synimmune/data/01_extGBmap/immune_subsets/CD4_CD8.h5ad")
cd4_cd8_filtered = filter_donors(cd4_cd8)

tam_bdm = sc.read_h5ad("/net/bq-storage/ag-cherrmann/bq_cinac/projects/synimmune/data/01_extGBmap/immune_subsets/TAM-BDM.h5ad")
tam_bdm_filtered = filter_donors(tam_bdm)

tam_mg = sc.read_h5ad("/net/bq-storage/ag-cherrmann/bq_cinac/projects/synimmune/data/01_extGBmap/immune_subsets/TAM-MG.h5ad")
tam_mg_filtered = filter_donors(tam_mg)
```

```python
cd4_cd8_filtered.write_h5ad("/net/bq-storage/ag-cherrmann/bq_cinac/projects/synimmune/data/01_extGBmap/immune_subsets/CD4_CD8_filtered.h5ad")
tam_bdm_filtered.write_h5ad("/net/bq-storage/ag-cherrmann/bq_cinac/projects/synimmune/data/01_extGBmap/immune_subsets/TAM-BDM_filtered.h5ad")
tam_mg_filtered.write_h5ad("/net/bq-storage/ag-cherrmann/bq_cinac/projects/synimmune/data/01_extGBmap/immune_subsets/TAM-MG_filtered.h5ad")
```
